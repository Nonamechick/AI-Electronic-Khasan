```python
# Required libraries
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
from sklearn import tree
import matplotlib.pyplot as plt

# 1. Create data
data = {
    'Income': [60000, 55000, 70000, 50000, 52000,
               48000, 45000, 40000, 30000, 35000,
               60000, 52000, 70000, 48000, 45000,
               50000, 51000, 49000, 53000, 55000],
    'Age':    [32, 35, 40, 30, 33,
               28, 25, 22, 27, 20,
               25, 28, 29, 30, 31,
               25, 29, 35, 31, 30],
    'Buy':    ['Yes', 'Yes', 'Yes', 'Yes', 'Yes',
               'No', 'No', 'No', 'No', 'No',
               'No', 'No', 'No', 'No', 'No',
               'No', 'No', 'No', 'Yes', 'Yes']
}

# Create DataFrame
df = pd.DataFrame(data)

# 2. Print DataFrame
print("--- DataFrame Output ---")
print(df)
```

```python
# Scatter plot
plt.figure(figsize=(8,6))

# Buy = Yes
yes = df[df['Buy'] == 'Yes']
plt.scatter(yes['Income'], yes['Age'], color='green', label='Buy: Yes', s=100, marker='o')

# Buy = No
no = df[df['Buy'] == 'No']
plt.scatter(no['Income'], no['Age'], color='red', label='Buy: No', s=100, marker='x')

# 그래프 설정
plt.xlabel('Income')
plt.ylabel('Age')
plt.title('Income vs Age Scatter Plot (Buy Yes/No)')
plt.legend()
plt.grid(True)
plt.show()
```
