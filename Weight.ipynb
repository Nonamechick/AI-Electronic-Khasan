```python
from sklearn import linear_model
regr = linear_model.LinearRegression()
X = [[168, 0], [166, 0], [173, 0], [165, 0], [177, 0], [163, 0], \
     [178, 0], [172, 0], [163, 1], [162, 1], [171, 1], [162, 1], \
     [164, 1], [162, 1], [158, 1], [173, 1], ]    #input data : 2 dimension
y = [65, 61, 68, 63, 68, 61, 76, 67, 55, 51, 59, 53, 61, 56, 44, 57] # y:1 dimension
regr.fit(X, y)         # learning
print('coef :', regr.coef_ )
print(' intercept :', regr.intercept_)
print(' regr_score :', regr.score(X, y))
print(' Estimated weight of mamur and sevinch  Khasan:', regr.predict([[167, 0], [167, 1], [175,0]]))
```
